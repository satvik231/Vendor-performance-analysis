```python
import pandas as pd
import os
from sqlalchemy import create_engine
```

```python
engine = create_engine('sqlite:///inventory.db')
```

```python
def ingest_db(df,table_name,engine,first_chunk):
    df.to_sql(
        table_name,
        con = engine,
        if_exists = 'replace' if first_chunk else 'append', index = False)
```

```python
for file in os.listdir('data'):
    if file.endswith('.csv'):
        table_name = file[:-4]

        total_rows = 0

        for i, chunk in enumerate(
            pd.read_csv('data/' + file, chunksize=100000)
        ):
            total_rows += len(chunk)
            print(f"{file} - Chunk {i}: {chunk.shape}")

            ingest_db(
                chunk,
                table_name,
                engine,
                first_chunk=(i == 0)
            )

        print(f"{file} loaded successfully. Total rows: {total_rows}")
```
